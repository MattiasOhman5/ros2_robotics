```python
import pandas as pd
import matplotlib.pyplot as plt

def plot(path):
    df = pd.read_csv(path, skiprows=1) 

    plt.figure(figsize=(8,4))
    plt.plot(df['state_size'], df['elapsed']*1000, 'o-', alpha=0.7)
    plt.xlabel('Number of landmarks')
    plt.ylabel('Computation time [ms]')
    plt.title('EKF-SLAM computation time vs landmarks')
    plt.grid(True)
    plt.show()
```

```python
path = "/home/mattias/ros2_robotics/src/r7021e_ekf_slam/datalogs/log_file_1.csv"
plot(path)
```
